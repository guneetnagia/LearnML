```python
from keras import layers
import keras
class MyModel(keras.Model):
  def __init__(self, num_classes=3, h1=8, h2=10, h3=10, **kwargs):
    super().__init__(**kwargs)
    self.l1 = layers.Dense(h1, activation='relu')
    self.l2 = layers.Dense(h2, activation='relu')
    self.l3 = layers.Dense(h3, activation='relu')
    self.output = layers.Dense(num_classes, activation='softmax')

  def call(self, inputs):
    x = self.l1(inputs)
    x = self.l2(x)
    x = self.l3(x)
    return self.out(x)
```
